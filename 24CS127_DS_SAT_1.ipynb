```python
import pandas as pd
```

```python
df = pd.read_csv('Q08_wildlife_camera_trap.csv')
```

```python
print(df.info())
print(df.isnull().sum())
```

```python
#remove 2 null values in trigger_type column
df = df.dropna(subset=['trigger_type'])
```

```python
print(df.isnull().sum())
```

```python
df = df[df['trigger_type'].str.lower() != 'blank']
```

```python
print(df.info())
print(df.isnull().sum())
```

All null values are removed

```python
df = df.drop_duplicates()
```

```python
df['sighting_time'] = pd.to_datetime(df['sighting_time'])
df['hour'] = df['sighting_time'].dt.hour
```

```python
# Species activity
print("Species Activity:")
print(df['species'].value_counts())
```

```python
# Species activity across locations
print("Species by Location:")
print(pd.crosstab(df['species'], df['location']))
```

```python
# Time of day
df['time_of_day'] = pd.cut(df['hour'],
    bins=[-1,5,11,17,23],
    labels=['Night','Morning','Afternoon','Evening'])
```

```python
print("Species by Time:")
print(pd.crosstab(df['species'], df['time_of_day']))
```

```python
# Nocturnal activity indicator
df['nocturnal'] = df['hour'].apply(
    lambda x: 'Yes' if x >= 18 or x < 6 else 'No')
```

```python
print("Nocturnal Activity:")
print(pd.crosstab(df['species'], df['nocturnal']))
```

```python
# Sighting frequency by location
print("Location Frequency:")
print(df['location'].value_counts())
```

```python
df.to_csv("cleaned_wildlife_camera_trap.csv", index=False)
```
